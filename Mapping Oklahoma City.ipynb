```python
import pandas as pd
import folium
```

```python
city_name = ["Norman", "Oklahoma City", "Edmond", "Stillwater"]
latitude = [35.2226, 35.4676, 35.6528, 36.1156]
longitude = [-97.4395, -97.5164, -97.4781, -97.0584]
description = ["ou community", "rough area", "rich man area", "non existent area"]
```

```python
places = pd.DataFrame({
    "city_name":city_name,
    "latitude":latitude,
        "longitude":longitude,
        "description":description  
})
```

```python
places
```

```python
my_map = folium.Map(location=[35.4676,-97.5164], zoom_start=10)
my_map
```

```python
for _, row in places.iterrows():
    folium.Marker(
        location=[row["latitude"], row["longitude"]],
        popup=row["city_name"],
        tooltip=row["city_name"]
    ).add_to(my_map)
```
