```python
import pandas as pd
import folium
```

```python
Norman = {
    "city_name": "Norman",
    "university_name": "University of Oklahoma",
    "population": 40000,
    "latitude": 35.2226,
    "longitude": -97.4395
}
```

```python
Edmond = {
    "city_name": "Edmond", 
    "university_name": "University of Central Oklahoma", 
    "population" : 7000, 
    "latitude": 35.6528, 
    "longitude":  -97.4781
}
```

```python
Stillwater = {
    "city_name": "Stillwater",
    "university_name": "Oklahoma State University",
    "population": 6900,
    "latitude": 36.1156,
    "longitude": -97.0584
}
```

```python
Tahlequah = {
    "city_name": "Tahlequah", 
    "university_name" : "Northeastern State University", 
    "population": 8670, 
    "latitude": 35.9151, 
    "longitude": -94.9700
}
```

```python
df = pd.DataFrame(
    [Norman, Stillwater, Edmond, Tahlequah]
)
```

```python
df
```

```python
my_map = folium.Map(location = [35.226, -97.4395],zoom_start = 7)
```

```python
my_map
```

```python
for _, row in df.iterrows(): 
    folium.Marker(
        location = [row["latitude"], row["longitude"]],
        popup = row["city_name"],
        tooltip = row["university_name"]
    ).add_to(my_map)
```

```python
my_map
```

```python
my_map.save("Oklahoma Universities.html")
```
